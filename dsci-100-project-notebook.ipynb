#### Task 3.1: Tidy the data (5 points)

Produce a tidy data set that could be used to answer your research question using the `tidyverse`. Please ensure you have at least one quantitative (numeric) and one categorical variable:
- Create new variables (if needed)
- Transform the data into a tidy form (if needed)
- Remove irrelevant columns (if needed)
- Comment your code throughout
- Show the first 6 rows of the tidied data.

</div>

```r
library(tidyverse)

# Select relative columns and create the new championship variable
champ_data <- March_madness |>
    # Only keep the variables needed for research question
    select(year, school, seed, reg_wins_pct, tourney_finish) |>
    # Remove rows without data
    filter(!is.na(seed), !is.na(reg_wins_pct)) |>
    # Remove special designation seed == 0 from 1983
    filter(seed != 0) |>
    # Create new variable that true if the team won the championship
    mutate(champion = as_factor(tourney_finish == "champ")) |>
    # Drop tourney_finish since champion replaces it
    select(year, school, seed, reg_wins_pct, champion)

# Show the first 6 rows of tidied data
slice(champ_data, 1:6)
```

#### Task 4.1: Calculate summary statistics of numeric variables (2 points)

- Use any functions from the `tidyverse` to create **one** table that outputs the mean, minimum, and maximum of **all** numeric columns in your data, dropping the missing values if they exist.
- Show the outputted table.

</div>

```r
# Start from tidied data
champ_data |>
    # Keep only numeric columns
    select(year, seed, reg_wins_pct) |>
    # Stack all selected columns into rows; old column names go into a "variable" column; their values go into a "value" column
    pivot_longer(everything(), names_to = "variable", values_to = "value") |>
    # Group rows by variables
    group_by(variable) |>
    # Calculate the statistics separately for each variable. Average, ignoring missing values; smallest value, ignoring missing values; largest value, ignoring missing values
    summarize(mean = mean(value, na.rm = TRUE), min = min(value, na.rm = TRUE), max = max(value, na.rm = TRUE))
```

#### Task 4.2: Determine the amount of missingness (2 points)

- Use your `dplyr` knowledge to create **one** table that calculates the number of missing values for each column in the data set (both numeric and categorical).
- Show the outputted table. 
</div>

```r
# Start from tidied data
champ_data |>
    # Apply the same function to every column; is.na() marks missing values, sum() counts them
    summarize(across(everything(), ~ sum(is.na(.x)))) |>
    #Turn the one-row result into a long table; column names go into "column"; the counts go into "n_missing"
    pivot_longer(everything(), names_to = "column", values_to = "n_missing")
```

```r
# Set the data and map win percentage to x-axis
ggplot(champ_data, aes(x = reg_wins_pct)) +
    # Each bar covers 5 percentage points; bar color fixed, so outside aes; white bar outlines for readability 
    geom_histogram(binwidth = 5, fill = "steelblue", color = "white") +
    # Plot title; x-axis label; y-axis label 
    labs(title = "Distribution of Regular Season Win Percentage", x = "Regular Season Win Percentage (%)", y = "Number of Team Seasons")
```

```r
# Start from tidied data
champ_data |>
    # Separate champions from non-champions
    group_by(champion) |>
    # Count the rows in each group
    summarize(count = n()) |>
    # Divide each count by the total to get a proportion
    mutate(proportion = count / sum(count)) |>
    # Map champion to x and proportion to y
    ggplot(aes(x = champion, y = proportion)) +
    # Use the proportion values as bar heights; bar color
    geom_bar(stat = "identity", fill = "steelblue") +
    # Plot title; x-axis label; y-axis label
    labs(title = "Proportion of Tournament Teams That Won Championship", x = "Won Championship (TRUE = yes)", y = "Proportion of Team Seasons")
```

Visualization 1: This density plot compares regular season win percentage for champions and non-champions. Non-champions peak around 75%, while champions are concentrated between about 85% and 100%, with almost none below 65%. The two groups overlap somewhat in the 80-90% range, but win percentage clearly separates champions from other teams, so it is a useful predictor for the K-NN model.

```r
# Win percentage on the x-axis; separate curve color for each champion group
ggplot(champ_data, aes(x = reg_wins_pct, color = champion)) +
    # Smooth curve showing each group's distribution
    geom_density() +
    # Plot title; x-axis label; y-axis label; legend title
    labs(title = "Regular Season Win Percentage by Tournament Outcome", x = "Regular Season Win Percentage (%)", y = "Density", color = "Champion")
```

Visualization 2: This bar chart shows the percentage of teams at each seed that won the championship. About 19% of 1 seeds won, compared with about 5% of 2 seeds, about 1% of 3 seeds, and essentially none of the lower seeds. Seed therefore carries strong information about the outcome, but the relationship is not a straight line and is concentrated in the top three seeds, which suits K-NN because it makes no linearity assumption.

```r
# Start from tidied data
champ_data |>
    # Calculate separately for each seed
    group_by(seed) |>
    # TRUE counts as 1, so the mean is the share of champions, times 100 gives a percent
    summarize(pct_champ = 100 * mean(champion == "TRUE")) |>
    # Seed on x, championship percent on y
    ggplot(aes(x = seed, y = pct_champ)) +
    # Use the computed percentages as bar heights; bar color
    geom_bar(stat = "identity", fill = "steelblue") +
    # Plot title; x-axis label; y-axis label
    labs(title = "Championship Rate by Tournament Seed", x = "Seed (1 = Highest)", y = "Percent of Teams Winning Championship")
```

I plan to use K-nearest neighbors classification, with seed and regular season win percentage as predictors and champion (yes/no) as the outcome. K-NN is appropriate because the outcome is categorical, and the method predicts a class by letting the k closest training teams vote. Its limitations are that it is sensitive to predictor scales, so I will center and scale both predictors in a recipe, and that only about 1.8% of team seasons are champions, so most neighbors will be non-champions and the model may almost never predict a champion. K-NN also gives no simple interpretation of how each predictor matters. I will use set.seed() for reproducibility and split the tidied data once into 75% training and 25% testing sets with initial_split(), stratified by champion so both sets keep the same class balance. On the training set I will use 5-fold stratified cross-validation (vfold_cv) with tune_grid to choose the number of neighbors k, so a separate validation set is not needed. The recipe will be learned on training data only, and the test set will be used once to evaluate the final model. Accuracy alone would be misleading because always predicting "not champion" is already about 98.2% accurate.
